# Manifesto File Loader

## Purpose

This notebook prepares manifesto files for use in the subsequent analysis pipeline.

The original manifesto files are identified using party IDs and election years. This notebook provides a utility to rename these files using readable party abbreviations.

For example:

```text
<party_id>_<date>.csv
        ↓
<party_abreviation>_<year>.csv
```

This produces filenames that are easier to use and interpret throughout the rest of the project.

## Pipeline

The data preparation performed in this notebook follows these steps:

1. Identify manifesto files in the target directory.
2. Extract the party ID and election date from each filename.
3. Match the party ID to the corresponding party information.
4. Extract the party abbreviation.
5. Convert the original date to the election year.
6. Rename the file using the standardized naming convention.
7. Leave files that do not follow the expected naming convention unchanged.

The resulting files are used as inputs by the subsequent analysis notebooks.


## 1. File renaming function

The function below implements the filename standardization described above.

It operates directly on the filenames and does not modify the contents of the CSV files.

Files that do not match the expected naming convention are ignored. This prevents already-processed files or unrelated files from being renamed accidentally.


````python
from pathlib import Path


def rename_party_files(directory: str, party_dict: dict) -> None:
    """
    Rename manifesto files using standardized party abbreviations.

    Expected input filename:
        PARTY_ID_DATE.csv

    Output filename:
        PARTY_ABBREVIATION_YEAR.csv

    Parameters
    ----------
    directory : str
        Directory containing the manifesto files.
    party_dict : dict
        Mapping from Manifesto Project party IDs to party information.
    """

    directory = Path(directory)

    for file in directory.glob("*.csv"):

        # Extract the filename without its extension.
        stem = file.stem

        # Ignore files that do not follow the expected naming convention.
        if "_" not in stem:
            continue

        party_id, date = stem.split("_", 1)

        # Convert the party ID to an integer for dictionary lookup.
        try:
            party_id = int(party_id)
        except ValueError:
            print(f"Ignored (non-numeric party ID): {file.name}")
            continue

        # Ignore files for which no party mapping is available.
        if party_id not in party_dict:
            print(f"Ignored (unknown party ID): {file.name}")
            continue

        abbreviation = party_dict[party_id]

        # Extract the election year from the original date.
        year = date[:4]

        new_name = f"{abbreviation}_{year}{file.suffix}"
        new_path = directory / new_name

        file.rename(new_path)

        print(f"[RENAMED] {file.name} -> {new_name}")
````

# Exemple d'utilisation

## 2. Input data

The manifesto files used by this notebook are stored in:

```text
data/raw/manifestos/usa/
```

The directory is expected to contain files whose names follow the naming convention provided by the original data source.
Along this project, if you want to add a new country you should create a folder with the country name and then download 
the manifestos from the website into the folder of the country. Then you should execute this notebook by setting the correct path
and the correct party mapper dictionary. An example is available at the end of the notebook.

The mapping below associates Manifesto Project party IDs with the abbreviations used in this project.


````python
usa_directory = "../data/raw/manifestos/usa"

party_mapper = {
    61320: ("Democratic Party", "dem"),
    61620: ("Republican Party", "rep")
} # Permet d'associer les id du site du Manifesto Project aux noms des partis
````

## 3. Standardize filenames

The function is now applied to the US manifesto corpus.

If the files have already been renamed, they will not be modified again because they no longer match the expected original filename format.

````python
rename_party_files(
    directory=usa_directory,
    party_dict=party_mapper
)
````
